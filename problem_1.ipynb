### Q1: Design & Commercialization Considerations

* **Aspect Ratio & Reach Constraints:** The crane features a compact vertical mounting height relative to its horizontal outreach. ### Q2: Material Selection Framework
**Question:** How should the material properties differ when selecting stock for the main structural arm members versus the connection hardware pins? Which material metrics (e.g., ductility, yield strength, shear strength) are prioritized for each, and why?

#### **Answer Key & Evaluation Rubric:**
* **Structural Arms (Beams $AC$ and $BC$):** These components require structural materials optimized for a high **stiffness-to-weight ratio** and high tensile **yield strength ($\sigma_y$)**. Low-to-medium carbon structural steels (such as A36) or structural aluminum alloys (such as 6061-T6) are standard choices. These materials provide high ductility, which ensures that if an unexpected overload occurs, the structural arms will bend visibly and absorb energy rather than suffering sudden, brittle fracture.
* **Connection Fastening Pins (Joints $A, B, C, D, E$):** Pins operate in high-wear rotational locations and must withstand heavy transverse cutting actions over very small cross-sectional areas. Therefore, their material selection prioritizes exceptionally high **shear yield strength ($\tau_y$)** and surface hardness. High-strength alloy steels (such as medium-carbon quenched and tempered steels) are utilized so that the pins maintain precise geometric clearances and resist deformation or surface scoring during service.

### Q3: Critical Geometrical Parameters
**Question:** Identify the critical geometric parameters of the system. Which specific dimensions govern the internal axial forces, which ones dictate buckling vulnerabilities, and which parameters define the localized shear failure planes of the hardware?

#### **Answer Key & Evaluation Rubric:**
* **Internal Axial Forces:** The **vertical wall-mount spacing ($H_{AB}$)** and the **triangulated length of the diagonal strut ($BC$)** are the primary parameters that govern internal member forces. As the vertical wall-mount spacing decreases, the horizontal components of the reaction forces multiply rapidly, significantly increasing the internal tension in horizontal beam $AC$ and the compression in diagonal strut $BC$.
* **Buckling & Flexural Stability:** The unsupported lengths of the individual segments ($AE, EC, BD, DC$) dictate the global structural stability. Because diagonal strut $BC$ experiences heavy internal compression, its total length and **cross-sectional radius of gyration ($r$)** are the critical geometric parameters governing its vulnerability to global elastic column buckling.
* **Hardware Shear Planes:** The **nominal diameter ($d$) of the pin** defines the localized failure plane. Because the wall bracket at Joint $A$ uses a double-shear clevis configuration, the total boundary reaction force is split evenly across exactly **two parallel failure cross-sections** ($A_{\text{pin}} = \frac{\pi d^2}{4}$).

### Q4: Loading Characteristics & Estimation
**Question:** What kind of loading is this system subjected to in an industrial environment? Is the loading strictly static, or is it time-varying/cyclic? How do dynamic field factors affect your estimation of the design load compared to the dead weight of the payload?

#### **Answer Key & Evaluation Rubric:**
* **Time-Varying & Cyclic Classification:** In a real factory environment, the crane is subjected to **time-varying and cyclic loading**. The structural elements undergo a discrete load cycle every single time a payload is hitched, hoisted, swung across the floor, and unhooked. Over thousands of cycles, this repetitive behavior introduces fatigue considerations.
* **Dynamic Shock Amplification:** The load cannot be modeled simply as a static dead weight ($600\text{ lb}$). Sudden hoisting acceleration, quick braking, or trolley jerking inject significant inertial forces into the structure. To account for this dynamically active environment, engineers scale up the static payload weight using a dynamic amplification factor or impact multiplier ($\mathcal{I}$):
  $$P_{\text{design}} = \mathcal{I} \cdot P_{\text{static}}$$
  For standard industrial electric hoists, a structural impact modifier of $\mathcal{I} = 1.15$ to $1.25$ ($15\%$ to $25\%$ overhead adjustment) is typically applied to convert dynamic effects into a conservative equivalent-static design load.

### Q5: Location Vulnerabilities, Strength, & Stiffness
**Question:** Where are the critical, highest-stress locations most vulnerable to failure? Furthermore, explain why the horizontal arm must be optimized for **stiffness** (deflection limits) even if it already possesses more than enough **strength** to prevent permanent failure.

#### **Answer Key & Evaluation Rubric:**
* **Critical Locations:** The highest stress concentrations occur at the primary wall interfaces—**Joint $A$ and Joint $B$**—where the entire cantilevered moment of the hanging payload must be resolved back into the rigid concrete pillar. Locally, the pin-holes themselves act as high geometric stress concentrations ($K_t$), making these joints highly vulnerable to localized metal tear-out, bearing deformation, or pin shear.
* **The Importance of Stiffness Optimization:** While a component must have sufficient **strength** to avoid fracturing or permanently yielding, crane components are frequently sized based on strict **stiffness criteria**. If the horizontal beam $AC$ lacks adequate flexural stiffness, the outermost tip (Point $C$) will sag excessively under load. This elastic deflection creates an unlevel track, causing the hanging trolley and its heavy payload to drift uncontrollably due to gravity, which ruins positioning accuracy and presents an extreme workspace hazard to ground operators.

### Q6: Evaluation Criteria & Structural Metrics
**Question:** What distinct numerical failure criteria and material limits should you apply to evaluate the main structural arms versus the fast-acting connector pins? Should your structural safety factor be benchmarked against the material's **yield strength ($\sigma_y$)** or its **ultimate tensile strength ($\sigma_{ut}$)**? Why?

#### **Answer Key & Evaluation Rubric:**
* **Evaluation Criteria for Fastening Pins:** The pins must be evaluated using a **shear yield criterion** (such as the Maximum Shear Stress/Tresca theory or Distortion Energy/von Mises theory) compared against the material's **shear yield strength ($\tau_y \approx 0.577\sigma_y$)**. Benchmarking against yielding is mandatory because any permanent bending or plastic gouging of a pin will permanently lock up the crane pivot mechanism and cause structural binding.
* **Evaluation Criteria for Structural Arms:** The structural arms are evaluated using the **von Mises equivalent stress criterion** compared directly against the tensile **yield strength ($\sigma_y$)** of the selected metal to establish a safe, linear-elastic operating window.
* **Why Yield Strength Over Ultimate Tensile Strength?** Sizing operational machinery based on the material's **ultimate tensile strength ($\sigma_{ut}$)** is an unsafe engineering practice. Long before a ductile metal beam reaches its ultimate breaking strength, it will undergo massive plastic deformation and permanent elongation. In a triangulated truss or crane assembly, this severe geometric distortion alters the intended load paths, accelerates destructive structural column buckling, and renders the machine completely inoperable. Sizing for yield strength guarantees safe, completely recoverable elastic behavior over the lifespan of the equipment.

### Step 1: Material and Fastener Parameter Selection
Select your structural member material and the connection hardware grade from the dropdown menus below to populate your design metrics.

```python
# 1. Standard Materials Database
materials_db = {
    'A36 Structural Steel': {'sigma_y': 36.0, 'sigma_ut': 58.0, 'cost': 1.0},
    '6061-T6 Aluminum': {'sigma_y': 40.0, 'sigma_ut': 45.0, 'cost': 2.2},
    'AISI 1045 HR Steel': {'sigma_y': 45.0, 'sigma_ut': 82.0, 'cost': 1.6}
}

# 2. Rigorous SAE J429 Bolt Database
bolts_db = {
    'SAE Grade 2 (Low Carbon Steel)': {'tau_allow': 16.4},
    'SAE Grade 5 (Medium Carbon Tempered)': {'tau_allow': 26.5},
    'SAE Grade 8 (Alloy Steel, Quenched & Tempered)': {'tau_allow': 37.5}
}

# 3. Standard Unified Coarse (UNC) Fastener Size Mapping
# Maps standard size names to their exact minor tensile core area (sq inches)
standard_sizes = {
    '1/4\" - 20 UNC': {'d': 0.2500, 'area': 0.0318},
    '5/16\" - 18 UNC': {'d': 0.3125, 'area': 0.0524},
    '3/8\" - 16 UNC': {'d': 0.3750, 'area': 0.0775},
    '7/16\" - 14 UNC': {'d': 0.4375, 'area': 0.1063},
    '1/2\" - 13 UNC': {'d': 0.5000, 'area': 0.1419},
    '5/8\" - 11 UNC': {'d': 0.6250, 'area': 0.2260}
}

# UI Dropdowns enforcing standard procurement choices
mat_dropdown = widgets.Dropdown(options=list(materials_db.keys()), description='Beam Material:', style={'description_width': 'initial'})
bolt_dropdown = widgets.Dropdown(options=list(bolts_db.keys()), description='Fastener Grade:', style={'description_width': 'initial'})
size_dropdown = widgets.Dropdown(options=list(standard_sizes.keys()), description='Standard Bolt Size:', style={'description_width': 'initial'})

ui = widgets.VBox([mat_dropdown, bolt_dropdown, size_dropdown])
display(ui)
```

### Step 2: Structural Verification and Stress Solver
Run this cell to calculate the global equilibrium state of the crane and check the safety factor of your pin connection at $A$ (under double shear conditions).

```python
def run_engineering_analysis(b):
    clear_output(wait=True)
    display(ui)
    display(calculate_btn)
    
    # ---- 1. FIXED SYSTEM PARAMETERS ----
    P = 600.0      # lb (Workspace Payload)
    L_AC = 4.0     # ft (Total horizontal arm length)
    L_AE = 2.0     # ft (Distance to mid-link joint E)
    H_AB = 3.0     # ft (Vertical distance between wall anchors)
    
    # ---- 2. FULL STRUCTURAL SOLVER (Rigorous Statics) ----
    F_Ax = (P * L_AC) / H_AB
    F_Ay = P / 3.0 
    F_A_total = np.sqrt(F_Ax**2 + F_Ay**2)
    
    # ---- 3. UPDATED PARAMETER RETRIEVAL FROM DROPDOWNS ----
    chosen_mat = mat_dropdown.value
    chosen_bolt = bolt_dropdown.value
    chosen_size = size_dropdown.value  # Grabs the standard string name (e.g., '3/8" - 16 UNC')
    
    mat_info = materials_db[chosen_mat]
    bolt_info = bolts_db[chosen_bolt]
    
    # Extract the standard nominal diameter and true core minor area from our database mapping
    d_pin = standard_sizes[chosen_size]['d']
    A_pin = standard_sizes[chosen_size]['area']  # Crucial: Using standard thread root area!
    
    # ---- 4. DOUBLE SHEAR ANALYSIS ----
    # Core shear force is distributed across 2 distinct thread/shank shear planes
    tau_avg_psi = F_A_total / (2.0 * A_pin)
    tau_avg_ksi = tau_avg_psi / 1000.0  
    
    tau_allow_bolt = bolt_info['tau_allow']
    f_s = tau_allow_bolt / tau_avg_ksi
    
    # ---- 5. REPORT GENERATION ----
    report = f"""
### ---- DESIGN EVALUATION REPORT ----
#### Global Statics Evaluation:
* Horizontal Reaction Vector ($F_{{Ax}}$): **{F_Ax:.1f} lb**
* Vertical Reaction Vector ($F_{{Ay}}$): **{F_Ay:.1f} lb**
* Resultant Force on Bracket Joint $A$ ($F_A$): **{F_A_total:.1f} lb** (True Vector Magnitude)

#### Pin Shear Analysis (Double Shear Standard Configurations):
* Selected Hardware Size: **{chosen_size}**
* Effective Bolt Diameter ($d$): **{d_pin:.4f} in.**
* Minor Tensile Core Area ($A_t$): **{A_pin:.4f} in²** (Thread Root Plane)
* Operating Average Shear Stress ($\\tau_{{avg}}$): **{tau_avg_ksi:.2f} ksi**
* Hardware Allowable Shear Strength ($\\tau_{{allow}}$): **{tau_allow_bolt:.2f} ksi**
"""
    display(Markdown(report))
    
    if f_s >= 1.5:
        status = f"### <span style='color:green'>DESIGN APPROVED</span>\n* **Factor of Safety (F.S.)**: {f_s:.2f}\n* **Economic Cost Index**: {mat_info['cost']:.1f}x"
    elif f_s >= 1.0:
        status = f"### <span style='color:orange'>WARNING: MARGINAL DESIGN</span>\n* **Factor of Safety (F.S.)**: {f_s:.2f} (Fails to meet industrial safety standard F.S. >= 1.5)\n* Change to high-strength pin material or select a larger standard thread size."
    else:
        status = f"### <span style='color:red'>CRITICAL FAILURE: DESIGN REJECTED</span>\n* **Factor of Safety (F.S.)**: {f_s:.2f} (Pin will undergo thread shear failure!)"
        
    display(Markdown(status))

# Re-register and display the layout button
calculate_btn = widgets.Button(description="Run Structural Solver", button_style='primary')
calculate_btn.on_click(run_engineering_analysis)
display(calculate_btn)
```

# Phase 3: Decision & Modeling Solutions & Grading Rubric

### 1. The $0.25\text{ in.}$ Baseline Test

#### **Analytical Calculation Verification:**
* **Resultant Boundary Force ($F_A$):** $$F_{Ax} = \frac{600\text{ lb} \cdot 4\text{ ft}}{3\text{ ft}} = 800.0\text{ lb}$$
  $$F_{Ay} = \frac{600\text{ lb}}{3} = 200.0\text{ lb}$$
  $$F_A = \sqrt{800^2 + 200^2} = 824.62\text{ lb}$$
* **Pin Shear Area ($A_{\text{pin}}$) under Double Shear:**
  $$A_{\text{pin}} = \frac{\pi}{4} (0.25\text{ in.})^2 = 0.04909\text{ in}^2$$
  $$\text{Total Shear Area} = 2 \cdot A_{\text{pin}} = 0.09817\text{ in}^2$$
* **Operating Average Shear Stress ($\tau_{\text{avg}}$):**
  $$\tau_{\text{avg}} = \frac{824.62\text{ lb}}{0.09817\text{ in}^2} = 8399.6\text{ psi} \approx 8.40\text{ ksi}$$
* **Factor of Safety ($F.S.$) for SAE Grade 2 Steel ($\tau_{\text{allow}} = 10.0\text{ ksi}$):**
  $$F.S. = \frac{\tau_{\text{allow}}}{\tau_{\text{avg}}} = \frac{10.0\text{ ksi}}{8.40\text{ ksi}} = 1.19$$

#### **Engineering Conclusion:**
**No, the baseline design is completely unacceptable for industrial deployment.** While the factor of safety is technically above $1.0$ ($F.S. = 1.19$), meaning it will not undergo immediate static failure under a perfectly quiescent load, it completely fails to meet standard industrial engineering safety mandates (typically requiring an absolute minimum margin of $F.S. \ge 1.5$ to $2.0$). 

Operating at an $F.S.$ of $1.19$ leaves virtually zero buffer for structural fatigue, cyclic degradation, or the dynamic shock amplification factors (such as hoisting inertial spikes) analyzed during Phase 1.

### 2. Economic Optimization Tradeoff: Path A vs. Path B

To resolve this structural vulnerability, students run the code to compare the two options:

#### **Path A: Increase Pin Hole Diameter to $0.375\text{ in.}$ (Maintaining SAE Grade 2 Steel)**
* **Geometric Change:** Sizing up to a standard $d = 0.375\text{ in.}$ bolt increases the single cross-sectional shear area to $A_{\text{pin}} = 0.1104\text{ in}^2$.
* **Resulting Stress:** $\tau_{\text{avg}} = \frac{824.62\text{ lb}}{2 \cdot 0.1104\text{ in}^2} = 3734.7\text{ psi} \approx 3.73\text{ ksi}$.
* **Resulting Factor of Safety:** $F.S. = \frac{10.0\text{ ksi}}{3.73\text{ ksi}} = \mathbf{2.68}$
* **Tradeoff Analysis:** This path yields an excellent, ultra-conservative safety factor well clear of standard industrial thresholds. However, it incurs significant **manufacturing labor overhead**. Machining and widening existing hardened structural wall brackets to accommodate a larger pin size requires manual labor, tooling wear, and line downtime. Furthermore, removing metal from the bracket ears reduces the net area available to resist localized bearing stresses and block shear tear-out around the hole boundary.

#### **Path B: Maintain $0.25\text{ in.}$ Pin Diameter (Upgrading to High-Strength SAE Grade 8 Steel)**
* **Material Change:** Upgrading the pin chemistry and heat-treatment profile jumps the allowable material shear limit to $\tau_{\text{allow}} = 25.0\text{ ksi}$.
* **Resulting Stress:** Stays at the baseline value of $\tau_{\text{avg}} = \mathbf{8.40\text{ ksi}}$.
* **Resulting Factor of Safety:** $F.S. = \frac{25.0\text{ ksi}}{8.40\text{ ksi}} = \mathbf{2.98}$
* **Tradeoff Analysis:** This configuration yields a premium factor of safety with zero modification to the bracket geometry. From an assembly standpoint, it offers an optimal **modular drop-in solution**. It eliminates the labor costs of redrilling the bracket ears and preserves the structural bulk of the mounting frame. The sole downside is a slightly higher raw procurement cost for individual specialized Grade 8 fasteners.

#### **Final Engineering Recommendation:**
**Path B is the superior engineering decision.** In high-throughput industrial settings like automotive assembly lines, minimizing manufacturing operations (drilling, deburring, and quality inspections) on structural assemblies drastically decreases lifecycle production expenses. 

While a Grade 8 bolt is more expensive on a component level than a Grade 2 bolt, the direct labor savings achieved by avoiding field modifications to the brackets far outweigh the raw material premium. Additionally, Path B maximizes the localized strength of the wall bracket housing by keeping the pin hole tightly clustered at $0.25\text{ in.}$, maintaining ideal edge-distance ratios ($e/d$) to prevent bearing tear-out failure under time-varying workplace conditions.
